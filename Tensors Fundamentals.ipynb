```python
#Aggregation in Tensors

x= torch.arange(0,100,10)

#Finding the maximum
print(f"Maximum: {x.max()}")

#Finding the minimum
print(f"Minimum: {x.min()}")

#Finding the mean
print(f"Mean: {x.type(torch.float32).mean()}")

#Finding the sum
print(f"Sum: {x.sum()}")
```

```python
#Reshaping, Stacking, Squeezing and Unsequezing

import torch

x = torch.arange(1.,8.)
x,x.shape
```

```python
x_reshaped = x.reshape(1,7)
x_reshaped,x_reshaped.shape
```

```python
#View example
z =x.view(1,7)
z,z.shape
```

```python
z[:,5] = 0
z,x
```

```python
#Stacke tensor on top of each other

x_stacked = torch.stack([x,x,x,x],dim=0)
x_stacked
```

```python
#Stacke tensor on top of each other

x_stacked = torch.stack([x,x,x,x],dim=1)
x_stacked
```

```python
#Squeezing

print(f"Previous tensor: {x_reshaped}")
print(f"Previour shape: {x_reshaped.shape}")

x_squeezed = x_reshaped.squeeze()
print(f"Squeezed tensor: {x_squeezed}")
print(f"Squeezed shape: {x_squeezed.shape}")
```

```python
#UnSqueezing

x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"Unsqueezed Tensor: {x_unsqueezed}")
print(f"Unsqueezed Shape: {x_unsqueezed.shape}")
```

```python
#Permuted

x_original = torch.rand(size=(6,7,8,9))

x_premuted = x_original.permute(3,2,0,1)

print(f"Original shape: {x_original.shape}")
print(f"Permuted shape: {x_premuted.shape}")
```

```python
#Pytorch Tensors and Numpy

#From Numpy array to tensors

import torch
import numpy as np

ndarray = np.arange(1.0,8.0)
tensor = torch.from_numpy(ndarray)
ndarray,tensor
```

```python
#Changing array, keeping the tensor same
ndarray = ndarray+1
ndarray,tensor
```

```python
#From tensor to numpy array

tensor = torch.arange(1.0,8.0)
ndarray = tensor.numpy()
tensor,ndarray
```

```python
#Changing Tensor, keeping array same

tensor=tensor+1
tensor,ndarray
```
